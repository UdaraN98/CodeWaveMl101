```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# import logisitc regression and Decision trree from sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# import shap and lime
import shap
from lime.lime_tabular import LimeTabularExplainer 


# Alternative approach to display LIME explanation
# Using direct HTML display to avoid IPython import issues
from IPython.display import HTML, display
```

```python
dataset = pd.read_csv('/Users/udaranilupul/Documents/Freelancing/CodeWave/CodeWaveMl101/Project/Data/customer_churn_dataset_prepared.csv')
```

```python
## Lime Explainer   for Logistic Regression
X = dataset.drop(columns=['Churn'])
y = dataset['Churn']    

model = LogisticRegression()
model.fit(X, y) 

explainer = LimeTabularExplainer(X.values, feature_names=X.columns, class_names=['No Churn', 'Churn'], discretize_continuous=True)  
i = 0  # Index of the instance to explain
exp = explainer.explain_instance(X.values[i], model.predict_proba, num_features=10)  
```

```python
# Alternative approach to display LIME explanation
# Direct HTML display to avoid IPython import issues in show_in_notebook()
from IPython.display import HTML, display

# Get the HTML representation directly and display it
html_explanation = exp.as_html(show_table=True)
display(HTML(html_explanation))
```
